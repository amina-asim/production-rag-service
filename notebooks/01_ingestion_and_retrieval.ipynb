```python
from pathlib import Path
from pypdf import PdfReader

# Find the PDF files
RAW_DIR = Path("../data/raw")
pdf_files = list(RAW_DIR.glob("*.pdf"))

print("PDFs found:", len(pdf_files))

# Extract text from each PDF
documents = []

for pdf_path in pdf_files:
    reader = PdfReader(str(pdf_path))

    for page_number, page in enumerate(reader.pages, start=1):
        text = page.extract_text() or ""

        if text.strip():
            documents.append({
                "text": text,
                "metadata": {
                    "filename": pdf_path.name,
                    "page": page_number
                }
            })

print("Pages extracted:", len(documents))

# Display previews
for doc in documents[:3]:
    print("\nFilename:", doc["metadata"]["filename"])
    print("Page:", doc["metadata"]["page"])
    print("Text preview:", doc["text"][:300])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# Configure the text splitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=150,
    separators=["\n\n", "\n", ". ", " ", ""]
)

chunks = []

for doc in documents:
    split_texts = text_splitter.split_text(doc["text"])

    for position, chunk_text in enumerate(split_texts):
        chunks.append({
            "text": chunk_text,
            "metadata": {
                "filename": doc["metadata"]["filename"],
                "page": doc["metadata"]["page"],
                "chunk_position": position
            }
        })

print("Total chunks created:", len(chunks))

# Preview the first chunk
print("\nFirst chunk:")
print(chunks[0]["text"])

print("\nMetadata:")
print(chunks[0]["metadata"])
```

```python
from sentence_transformers import SentenceTransformer

# Load the embedding model
embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

# Extract chunk text
chunk_texts = [chunk["text"] for chunk in chunks]

# Convert chunks into numerical vectors
embeddings = embedding_model.encode(
    chunk_texts,
    show_progress_bar=True
)

print("Number of chunks:", len(chunk_texts))
print("Embedding shape:", embeddings.shape)
print("First embedding preview:", embeddings[0][:10])
```

```python
import chromadb
from pathlib import Path

# Create a persistent ChromaDB database
CHROMA_DIR = Path("../data/chroma_db")
CHROMA_DIR.mkdir(parents=True, exist_ok=True)

client = chromadb.PersistentClient(
    path=str(CHROMA_DIR)
)

# Create or access our collection
collection = client.get_or_create_collection(
    name="rag_documents",
    metadata={"hnsw:space": "cosine"}
)

# Store chunks, embeddings, and metadata
collection.upsert(
    ids=[f"chunk_{i}" for i in range(len(chunks))],
    documents=[chunk["text"] for chunk in chunks],
    embeddings=embeddings.tolist(),
    metadatas=[chunk["metadata"] for chunk in chunks]
)

print("Documents stored successfully!")
print("Total records in database:", collection.count())
```

```python
# Write a question related to your PDFs
query = "What are the main topics discussed in these documents?"

# Convert the question into an embedding
query_embedding = embedding_model.encode(query).tolist()

# Search ChromaDB for the 3 most relevant chunks
results = collection.query(
    query_embeddings=[query_embedding],
    n_results=3
)

# Display retrieved results
for i in range(len(results["documents"][0])):
    print(f"\n--- Result {i+1} ---")
    print("Text:", results["documents"][0][i])
    print("Metadata:", results["metadatas"][0][i])
    print("Distance:", results["distances"][0][i])
```

```python
def retrieve_documents(query, k=4):
    # Convert question into an embedding
    query_embedding = embedding_model.encode(query).tolist()

    # Search ChromaDB
    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=k
    )

    retrieved_chunks = []

    for i in range(len(results["documents"][0])):
        retrieved_chunks.append({
            "text": results["documents"][0][i],
            "filename": results["metadatas"][0][i]["filename"],
            "page": results["metadatas"][0][i]["page"],
            "distance": results["distances"][0][i]
        })

    return retrieved_chunks


# Test the function
question = "What information is available in these documents?"

retrieved = retrieve_documents(question, k=4)

for i, item in enumerate(retrieved, start=1):
    print(f"\n--- Retrieved Chunk {i} ---")
    print("Source:", item["filename"])
    print("Page:", item["page"])
    print("Distance:", round(item["distance"], 4))
    print("Text:", item["text"][:300])
```

```python
def rag_answer(question, k=4):
    
    # 1. Retrieve relevant chunks
    retrieved_chunks = retrieve_documents(question, k=k)

    # 2. Prepare context
    context = build_context(retrieved_chunks)

    # 3. Create instructions for Gemini
    prompt = f"""
You are a document question-answering assistant.

Answer the user's question using ONLY the information
provided in the context below.

Rules:
1. Do not use outside knowledge.
2. Do not invent facts.
3. If the context does not contain the answer,
   respond exactly:
   "I cannot find the answer in the provided documents."
4. Include citations using the filename and page number.
5. Keep your answer clear and concise.

CONTEXT:
{context}

USER QUESTION:
{question}

ANSWER:
"""

    # 4. Generate answer
    response = gemini_client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return {
        "answer": response.text,
        "sources": retrieved_chunks
    }


print("RAG answer function created successfully!")
```

```python
def rag_answer(question, k=4):
    
    retrieved_chunks = retrieve_documents(question, k=k)

    # Find the best (lowest) distance
    best_distance = min(
        chunk["distance"] for chunk in retrieved_chunks
    )

    # Reject questions with irrelevant retrieved results
    if best_distance > 0.50:
        return {
            "answer": "I cannot find the answer in the provided documents.",
            "sources": []
        }

    context = build_context(retrieved_chunks)

    prompt = f"""
You are a document question-answering assistant.

Answer using ONLY the supplied context.

Rules:
1. Never use outside knowledge.
2. Never invent facts.
3. If the context does not answer the question,
   say: I cannot find the answer in the provided documents.
4. Cite sources using ONLY the supplied Source IDs.
5. Every factual claim must have a citation.
6. Do not invent filenames or page numbers.

CONTEXT:
{context}

QUESTION:
{question}

ANSWER:
"""

    response = gemini_client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return {
        "answer": response.text,
        "sources": retrieved_chunks
    }
```

```python
import re

def validate_citations(answer, sources):

    cited_ids = re.findall(
        r"\[Source ID: (\d+)\]",
        answer
    )

    valid_ids = set(
        str(i) for i in range(1, len(sources) + 1)
    )

    invalid_ids = [
        source_id
        for source_id in cited_ids
        if source_id not in valid_ids
    ]

    return {
        "cited_ids": cited_ids,
        "invalid_ids": invalid_ids,
        "all_valid": len(invalid_ids) == 0
    }
```

```python
result = rag_answer(
    "Which algorithms are used for word embeddings?"
)

citation_check = validate_citations(
    result["answer"],
    result["sources"]
)

print("ANSWER:")
print(result["answer"])

print("\nCITATION VALIDATION:")
print(citation_check)
```
